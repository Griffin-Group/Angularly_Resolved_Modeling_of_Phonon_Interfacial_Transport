```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('sige.freq.gp')
cm_to_thz = 0.0299792458

for band in range(1, 97):
    plt.plot(data[:, 0], data[:, band] * cm_to_thz, linewidth=2, alpha=1, color='#C09748')

plt.plot(data[:, 0], data[:, 6] * cm_to_thz, linewidth=2, alpha=1, color='#C09748')

plt.axvline(x=data[0, 0], linewidth=0.5, color='r', alpha=0.5)
plt.axvline(x=data[100, 0], linewidth=0.5, color='k', alpha=0.5)
plt.axvline(x=data[200, 0], linewidth=0.5, color='k', alpha=0.5)
plt.axvline(x=data[300, 0], linewidth=0.5, color='k', alpha=0.5) 
plt.axvline(x=data[400, 0], linewidth=0.5, color='k', alpha=0.5)  
plt.axvline(x=data[500, 0], linewidth=0.5, color='k', alpha=0.5)
plt.axvline(x=data[600, 0], linewidth=0.5, color='k', alpha=0.5)
plt.axvline(x=data[700, 0], linewidth=0.5, color='k', alpha=0.5)



plt.xticks(ticks=[0, data[100, 0], data[200, 0], data[300, 0], data[400, 0], data[500, 0], data[600, 0], data[700, 0]],
           labels=['$\Gamma$', 'Z', 'D', 'B', '$\Gamma$', 'A', 'E', 'Z'],
           fontname='Arial', fontsize=16)

plt.axhline(y=0, color='black', linestyle='--', linewidth=1, alpha=0.7)
plt.ylabel("Frequency (THz)", fontname='Arial', fontsize=16)
plt.xlim(data[0, 0], data[-1, 0])
plt.ylim(-2, 500 * cm_to_thz)
plt.tick_params(axis='both', which='major', labelsize=12, width=2)
plt.tick_params(axis='both', which='minor', labelsize=10, width=2)
#plt.savefig('C2m_2026.png', dpi=400, bbox_inches='tight')
plt.show()
```
