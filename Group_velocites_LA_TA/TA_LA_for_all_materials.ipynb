## NbN

```python
import numpy as np
import matplotlib.pyplot as plt

INFILE = "NbN_P-43m_GX.txt"

data = np.loadtxt(INFILE)
dist, band1, band3 = data[:, 0], data[:, 1], data[:, 3]

# Linear fits: freq = slope * distance + intercept
slope1, intercept1 = np.polyfit(dist, band1, 1)
slope3, intercept3 = np.polyfit(dist, band3, 1)

print(f"Band1: slope = {slope1:.6f}, intercept = {intercept1:.6f}")
print(f"Band3: slope = {slope3:.6f}, intercept = {intercept3:.6f}")

# Plot - two side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 5))

ax1.plot(dist, band1, 'o', color='#1f77b4', markersize=4, label='Band 1 data')
ax1.plot(dist, slope1 * dist + intercept1, '-', color='#1f77b4', label=f'Fit (slope={slope1:.3f})')
ax1.set_xlabel("Distance ($\AA^{-1}$)", fontsize=13)
ax1.set_ylabel("Frequency (THz)", fontsize=13)
ax1.set_title("TA", fontsize=13)
ax1.legend(fontsize=10)
ax1.grid(alpha=0.3, linestyle=':')

ax2.plot(dist, band3, 's', color='#d62728', markersize=4, label='Band 3 data')
ax2.plot(dist, slope3 * dist + intercept3, '-', color='#d62728', label=f'Fit (slope={slope3:.3f})')
ax2.set_xlabel("Distance ($\AA^{-1}$)", fontsize=13)
ax2.set_ylabel("Frequency (THz)", fontsize=13)
ax2.set_title("LA", fontsize=13)
ax2.legend(fontsize=10)
ax2.grid(alpha=0.3, linestyle=':')

fig.tight_layout()
#fig.savefig("band_fits.png", dpi=300, bbox_inches="tight")
plt.show()
```
