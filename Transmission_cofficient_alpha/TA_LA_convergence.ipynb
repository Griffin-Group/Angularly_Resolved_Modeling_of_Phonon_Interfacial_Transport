## Al: G-X path; $[100]$ crystallographic direction

```python
import matplotlib.pyplot as plt

kpoints = ['6x6x6', '8x8x8', '10x10x10', '12x12x12']
ta_values = [3.653, 3.538, 3.523, 3.465]
la_values = [7.335, 6.933, 6.786, 6.862]

plt.figure(figsize=(6, 6))

plt.plot(kpoints, ta_values, marker='o', linestyle='-', color='blue', linewidth=2, markersize=8, label='TA Branch')
plt.plot(kpoints, la_values, marker='s', linestyle='-', color='red', linewidth=2, markersize=8, label='LA Branch')

plt.xlabel('K-point Mesh Size')
plt.ylabel('Speed of Sound (km/s)')
plt.title('Convergence of Speed of Sound with K-point Mesh Density of Al')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
