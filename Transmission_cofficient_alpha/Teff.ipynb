```python
all_Teff = {'SiGe/Al': (0.954, 0.889), 'SiGe/Nb': (0.889, 0.828), 'SiGe/NbN': (0.860, 0.546), 'SiGe/Si': (0.508, 0.549)}
```

```python
import matplotlib.pyplot as plt
materials = list(all_Teff.keys())
ta_values = [result[0] for result in all_Teff.values()]
la_values = [result[1] for result in all_Teff.values()]

x = np.arange(len(materials))
width = 0.35

fig, ax = plt.subplots(figsize=(6, 6))
bars1 = ax.bar(x - width/2, ta_values, width, label='TA', color='#16A0AC')
bars2 = ax.bar(x + width/2, la_values, width, label='LA', color='#B4E2E8')

ax.set_ylabel('Effective transmission', fontsize=16) # $T_{eff}$')
ax.tick_params(axis='y', labelsize=16)
#ax.set_title('Angularly-Averaged Transmission Probability')

ax.set_xticks(x)
arrow_labels = [m.replace('/', ' → ') for m in materials]   # <-- arrow labels
ax.set_xticklabels(arrow_labels, fontsize=16, rotation=45, ha='right')

#ax.set_xticks(x)
#ax.set_xticklabels(materials, fontsize=16)
ax.legend(loc='upper right')
#plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('Teff_july.png', dpi=400)
plt.show()
```
